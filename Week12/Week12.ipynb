```python
# NAME: rectangle.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS

def calculate_rectangle(length, width):

    # calculate area using length * width
    # calculate perimeter using 2 * (length + width)
    # pack both values into a tuple (area, perimeter)
    # return the tuple
    area = length * width
    perimeter = 2 * (length + width)
    return (area, perimeter)


# PROGRAM LOGIC

length = float(input("Enter length: "))
width = float(input("Enter width: "))

my_area, my_perimeter = calculate_rectangle(length, width)

print("--- Results ---")
print("The Area is:", my_area)
print("The Perimeter is:", my_perimeter)
```
