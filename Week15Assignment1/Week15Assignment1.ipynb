```python
import json
import os

TASKS_FILE = "tasks.json"


def load_tasks():
    if os.path.exists(TASKS_FILE):
        with open(TASKS_FILE, "r") as f:
            return json.load(f)
    return []


def save_tasks(tasks):
    with open(TASKS_FILE, "w") as f:
        json.dump(tasks, f, indent=4)


def add_task(tasks):
    name = input("Enter task name: ")
    priority = input("Enter priority (1=High, 2=Medium, 3=Low): ")

    if priority not in ["1", "2", "3"]:
        print("Invalid priority.")
        return

    tasks.append({"name": name, "priority": int(priority)})
    save_tasks(tasks)
    print("Task added!")


def view_tasks(tasks):
    if not tasks:
        print("No tasks available.")
        return

    for i, task in enumerate(tasks):
        print(f"{i + 1}. {task['name']} (Priority {task['priority']})")


def complete_task(tasks):
    view_tasks(tasks)
    try:
        index = int(input("Enter task number to complete: ")) - 1

        if 0 <= index < len(tasks):
            completed = tasks.pop(index)

            with open("history.txt", "a") as f:
                f.write(completed["name"] + "\n")

            save_tasks(tasks)
            print("Task completed and archived!")
        else:
            print("Invalid selection.")
    except ValueError:
        print("Invalid input. Please enter a number.")


def main():
    tasks = load_tasks()

    while True:
        print("\n1. Add Task")
        print("2. View Tasks")
        print("3. Complete Task")
        print("4. Exit")

        choice = input("Choose an option: ")

        if choice == "1":
            add_task(tasks)
        elif choice == "2":
            view_tasks(tasks)
        elif choice == "3":
            complete_task(tasks)
        elif choice == "4":
            break
        else:
            print("Invalid choice.")


if __name__ == "__main__":
    main()
```
