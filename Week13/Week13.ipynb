```python
# NAME: journal.py

#IMPORT STATEMENTS

# FUNCTION write_entry
# open file in append mode
# write entry + newline
# close file

# FUNCTION read_journal
# open file in read mode
# read all contents
# print contents
# close file

def write_entry(filename, entry):
    file = open(filename, 'a')
    file.write(entry + "\n")
    file.close()


# PROGRAM LOGIC

# set filename to "journal.txt"

# loop until user quits
    # ask user: Write, Read, or Quit?

    # if Write
        # ask for entry
        # call write_entry
        # print "Entry saved"

    # if Read
        # call read_journal

    # if Quit
        # stop loop

    # otherwise
        # print "Invalid choice"

filename = "journal.txt"

while True:
    choice = input("(W)rite, (R)ead, or (Q)uit? ").upper()

    if choice == "W":
        entry = input("Enter your entry: ")
        write_entry(filename, entry)
        print("Entry saved.\n")

    elif choice == "R":
        print("Reading feature coming soon...\n")   # placeholder for now

    elif choice == "Q":
        print("Goodbye!")
        break

    else:
        print("Invalid choice. Try again.\n")
```
