```python
# NAME: expense_tracker.py

# IMPORTS
# (none needed for now)

# FUNCTION DEFINITIONS

# Function to add an expense
# - Ask user for category, vendor, amount, date
# - Store data in a dictionary
# - Return the dictionary

# Function to get total of all expenses
# - Loop through list
# - Add all amounts

# Function to get total by category
# - Ask for category
# - Loop through list
# - If category matches, add amount

# Function to get total by vendor
# - Ask for vendor
# - Loop through list
# - If vendor matches, add amount

expenses = []

def add_expense():
    category = input("Enter Category: ")
    vendor = input("Enter Vendor: ")

    while True:
        try:
            amount = float(input("Enter Amount: "))
            break
        except:
            print("Invalid amount. Try again.")

    date = input("Enter Date: ")

    expense = {
        "category": category,
        "vendor": vendor,
        "amount": amount,
        "date": date
    }

    return expense


# PROGRAM LOGIC

# Create empty list to store expenses

# Start main loop
# Display menu:
#   1. Add Expense
#   2. Get Totals
#   3. Exit

while True:
    print("\n--- Expense Tracker ---")
    print("1. Add New Expense")
    print("2. Get Totals")
    print("3. Exit")

    choice = input("Selection: ")

    if choice == "1":
        exp = add_expense()
        expenses.append(exp)
        print("Expense added!")

    elif choice == "2":
        print(expenses)  # temporary test

    elif choice == "3":
        break

# If user selects 1:
#   Call add function
#   Add returned dictionary to list

# If user selects 2:
#   Show sub-menu:
#       A: All
#       C: Category
#       V: Vendor

# If user selects 3:
#   End program
```
