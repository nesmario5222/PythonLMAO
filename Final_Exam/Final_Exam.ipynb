```python
# NAME: expense_tracker.py

# IMPORTS
# (none needed for now)

# FUNCTION DEFINITIONS

# Function to add an expense
# - Ask user for category, vendor, amount, date
# - Store data in a dictionary
# - Return the dictionary

# Function to get total of all expenses
# - Loop through list
# - Add all amounts

# Function to get total by category
# - Ask for category
# - Loop through list
# - If category matches, add amount

# Function to get total by vendor
# - Ask for vendor
# - Loop through list
# - If vendor matches, add amount

expenses = []

def add_expense():
    category = input("Enter Category: ")
    vendor = input("Enter Vendor: ")

    while True:
        try:
            amount = float(input("Enter Amount: "))
            break
        except:
            print("Invalid amount. Try again.")

    date = input("Enter Date: ")

    expense = {
        "category": category,
        "vendor": vendor,
        "amount": amount,
        "date": date
    }

    return expense

def total_all(expenses):
    total = 0
    for exp in expenses:
        total += exp["amount"]
    return total

def total_by_category(expenses, category):
    total = 0
    for exp in expenses:
        if exp["category"].lower() == category.lower():
            total += exp["amount"]
    return total

def total_by_vendor(expenses, vendor):
    total = 0
    for exp in expenses:
        if exp["vendor"].lower() == vendor.lower():
            total += exp["amount"]
    return total


# PROGRAM LOGIC

# Create empty list to store expenses

# Start main loop
# Display menu:
#   1. Add Expense
#   2. Get Totals
#   3. Exit

while True:
    print("\n--- Expense Tracker ---")
    print("1. Add New Expense")
    print("2. Get Totals")
    print("3. Exit")

    choice = input("Selection: ")

    if choice == "1":
        exp = add_expense()
        expenses.append(exp)
        print("Expense added!")

    elif choice == "2":
        print("\n--- Totaling Options ---")
        print("A: All Expenses")
        print("C: Filter by Category")
        print("V: Filter by Vendor")

        option = input("Choice: ").upper()

        if option == "A":
            total = total_all(expenses)
            print(f"Total of all expenses: ${total:.2f}")

        elif option == "C":
            cat = input("Enter Category: ")
            total = total_by_category(expenses, cat)
            print(f"Total for {cat}: ${total:.2f}")

        elif option == "V":
            ven = input("Enter Vendor: ")
            total = total_by_vendor(expenses, ven)
            print(f"Total for {ven}: ${total:.2f}")

    elif choice == "3":
        break

# If user selects 1:
#   Call add function
#   Add returned dictionary to list

# If user selects 2:
#   Show sub-menu:
#       A: All
#       C: Category
#       V: Vendor

# If user selects 3:
#   End program
```
