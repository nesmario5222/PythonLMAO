```python
import json
import os

FILE = "expenses.json"

def load_data():
    if os.path.exists(FILE):
        with open(FILE, "r") as f:
            return json.load(f)
    return []

def save_data(data):
    with open(FILE, "w") as f:
        json.dump(data, f)

def add_expense(data):
    name = input("Expense name: ")
    amount = float(input("Amount: "))
    category = input("Category: ")

    data.append({
        "name": name,
        "amount": amount,
        "category": category
    })

    save_data(data)

def view_expenses(data):
    for e in data:
        print(f"{e['name']} - ${e['amount']} ({e['category']})")

def total_spending(data):
    total = sum(e["amount"] for e in data)
    print(f"Total spending: ${total}")

def filter_expenses_by_category(data):
    category_filter = input("Enter category to filter by: ").strip().lower()
    filtered_expenses = [e for e in data if e['category'].lower() == category_filter]
    if filtered_expenses:
        print(f"\nExpenses in category '{category_filter.capitalize()}':")
        for e in filtered_expenses:
            print(f"{e['name']} - ${e['amount']} ({e['category']})")
    else:
        print(f"No expenses found in category '{category_filter.capitalize()}', or category not found.")

def main():
    data = load_data()

    while True:
        print("\n1. Add Expense")
        print("2. View Expenses")
        print("3. Filter by Category")
        print("4. Total Spending")
        print("5. Exit")

        choice = input("Choice: ")

        if choice == "1":
            add_expense(data)
        elif choice == "2":
            view_expenses(data)
        elif choice == "3":
            filter_expenses_by_category(data)
        elif choice == "4":
            total_spending(data)
        elif choice == "5":
            break

main()
```
