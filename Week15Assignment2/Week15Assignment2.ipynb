```python
import json
import os

FILE = "expenses.json"

def load_data():
    if os.path.exists(FILE):
        with open(FILE, "r") as f:
            return json.load(f)
    return []

def save_data(data):
    with open(FILE, "w") as f:
        json.dump(data, f)

def add_expense(data):
    name = input("Expense name: ")
    amount = float(input("Amount: "))
    category = input("Category: ")

    data.append({
        "name": name,
        "amount": amount,
        "category": category
    })

    save_data(data)

def view_expenses(data):
    for e in data:
        print(f"{e['name']} - ${e['amount']} ({e['category']})")

def main():
    data = load_data()

    while True:
        print("\n1. Add Expense")
        print("2. View Expenses")
        print("3. Exit")

        choice = input("Choice: ")

        if choice == "1":
            add_expense(data)
        elif choice == "2":
            view_expenses(data)
        elif choice == "3":
            break

main()
```
