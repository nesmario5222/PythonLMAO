```python
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
# Define a function named is_palindrome that takes one parameter (word)
# Convert the word to lowercase to make the check case insensitive
# Reverse the word using string slicing
# Compare the lowercase word to its reversed version
# Return True if they match, otherwise return False
def is_palindrome(word):
    word = word.lower()
    reversed_word = word[::-1]
    return word == reversed_word

user_word = input("Enter a word to test if it is a palindrome: ")
result = is_palindrome(user_word)

if result:
    print(f"{user_word} is a palindrome!")
else:
  print(f"{user_word} is not a palindrome.")

# PROGRAM LOGIC
# Prompt the user to enter a word
# Call the is_palindrome function with the user's word
# Store the returned value
# If True, print a message saying it is a palindrome
# If False, print a message saying it is not a palindrome
```
