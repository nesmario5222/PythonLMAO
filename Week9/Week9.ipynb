```python
# NAME: rentalcar.py

def get_classification():
    while True:
        code = input("Enter classification code (B, S, L): ").upper()
        if code in ["B", "S", "L"]:
            return code
        else:
            print("Invalid code. Try again.")


def get_rental_data():
    start = int(input("Start Odometer: "))
    end = int(input("End Odometer: "))
    days = int(input("Days Rented: "))

    miles = end - start
    return miles, days


def calculate_bill(code, miles, days):

    charged_miles = 0

    if code == "B":
        total_bill = days * 40

    elif code == "S":
        free_miles = 100
        if miles > free_miles:
            charged_miles = miles - free_miles
        total_bill = (days * 60) + (charged_miles * 0.35)

    elif code == "L":
        free_miles = days * 100
        if miles > free_miles:
            charged_miles = miles - free_miles
        total_bill = (days * 100) + (charged_miles * 0.50)

    return charged_miles, total_bill


def print_summary(code, miles, charged_miles, total_bill):

    names = {
        "B": "Budget",
        "S": "Standard",
        "L": "Luxury"}

    print("--- RENTAL SUMMARY ---")
    print("Classification:", names[code])
    print("Total Miles:", miles)
    print("Charged Miles:", charged_miles)
    print("Total Bill: $", round(total_bill, 2))



# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
# calculate_bill(classification, miles, days)
#   determine rate and free miles
#   calculate charged miles
#   calculate total cost
#   return charged miles and total bill

# PROGRAM LOGIC
# ask user for classification code (B, S, L)
# ask for start odometer
# ask for end odometer
# ask for days rented
# calculate miles driven
# call calculate_bill function
# print rental summary

code = get_classification()

miles, days = get_rental_data()

charged_miles, total_bill = calculate_bill(code, miles, days)

# Call the print_summary function to display the full rental summary
print_summary(code, miles, charged_miles, total_bill)
```
