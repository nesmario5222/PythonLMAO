```python
# IMPORT STATEMENTS
# (None)

# FUNCTION DEFINITIONS
# 1. Define verify_pin():
def verify_pin():
    # keep asking until the user enters the correct PIN
    while True:
        pin = input("Enter your PIN: ")
        if pin == "1234":
            print("PIN verified.")
            return True
        else:
            print("Incorrect PIN. Try again.")
verify_pin()
#    - Create a loop that asks for a PIN until it equals "1234"
#    - Return True
# 2. Define withdraw(balance, amount):
def withdraw(balance, amount):
    if amount > balance:
        print("Insufficient funds.")
        return balance
    else:
        return balance - amount
#    - Check if enough money exists
#    - Return the new (or old) balance

# PROGRAM LOGIC
# 1. Set current_balance to 500
# 2. Call verify_pin()
# 3. Get withdrawal amount from user input
# 4. Update current_balance by calling withdraw()
# 5. Print the final balance
current_balance = 500
withdrawal_str = input("Enter amount to withdraw: $")
withdrawal_amount = float(withdrawal_str)
current_balance = withdraw(current_balance, withdrawal_amount)
print(f"Your final balance is: ${current_balance:}")
```
