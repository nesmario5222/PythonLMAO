```python
# NAME: roster.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS

def add_players(roster):
    while True:
        name = input("Enter player name (or 'quit' to stop): ")

        if name == "quit":
            break

        roster.append(name)

def is_on_roster(roster, search_name):
    for player in roster:
        if player == search_name:
            return True
    return False


# add_players(list)
# - loop forever
# - ask user for a name
# - if user types "quit", stop loop
# - otherwise append name to list

# is_on_roster(list, name)
# - loop through list
# - if name matches any item, return True
# - if loop ends, return False

# PROGRAM LOGIC
# - create empty list called roster
# - call add_players(roster)
# - ask user for a name to search
# - call is_on_roster(roster, name)
# - print result message

roster = []

add_players(roster)

search = input("Who would you like to search for? ")

if is_on_roster(roster, search):
    print("Result: Player found!")
else:
    print("Result: Player not found.")
```
